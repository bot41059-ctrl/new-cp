```python
# ============================================================
# 🚀 Step 3: Start Server & Generate Live Cloudflare Link
# ============================================================
import subprocess, time, urllib.request, sys

# Purana running process kill karein
!fuser -k 8000/tcp > /dev/null 2>&1 || true

log_file = open("/content/uvicorn.log", "w")
server = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", "app:app", "--host", "127.0.0.1", "--port", "8000"],
    stdout=log_file,
    stderr=log_file
)

print("⏳ Backend Server start ho raha hai... intezar karein...")
server_live = False
for i in range(25):
    time.sleep(1)
    if server.poll() is not None:
        print("❌ Server crash ho gaya! Log details:")
        with open("/content/uvicorn.log", "r") as f:
            print(f.read())
        raise SystemExit("Backend start nahi ho paya.")
    try:
        with urllib.request.urlopen("http://127.0.0.1:8000", timeout=1) as resp:
            server_live = True
            break
    except Exception:
        pass

if server_live:
    print("✅ Backend Server 100% LIVE hai (Port 8000)!")
    print("\n" + "="*65)
    print("🌐 AAPKI WEBSITE KA LIVE LINK NEECHE GENERATE HO RAHA HAI:")
    print("   (Link aane ke baad uspar click karein, website khul jayegi)")
    print("="*65 + "\n")
    !cloudflared tunnel --url http://127.0.0.1:8000
else:
    print("❌ Server timeout! Log content:")
    with open("/content/uvicorn.log", "r") as f:
        print(f.read())
```
